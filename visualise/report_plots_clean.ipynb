# Utilities

```python
def load_json_with_continuations(full_path):
    """Load a metrics json file, merging any '_s0'/'_s1'/'_s2' continuation files next to it.

    Logging for a run sometimes got cut off partway through training and the rest of the
    epochs were logged into a separate per-seed file named like '<base>_s0.json',
    '<base>_s1.json', '<base>_s2.json' (one continuation file per seed) instead of being
    appended to the original file. Each continuation file has the same top-level structure
    as the base file but its 'ft' list holds only the one seed's remaining stats entries.
    This merges those extra stats entries into the matching seed so no epochs are silently
    dropped, and prints what it did (or didn't find) so gaps are traceable to a specific file.
    """
    with open(full_path) as f:
        data = json.load(f)

    base_no_ext, ext = os.path.splitext(full_path)
    continuation_paths = sorted(glob.glob(f"{base_no_ext}_s[0-9]*{ext}"))
    if not continuation_paths:
        return data

    print(f"Found {len(continuation_paths)} continuation file(s) for {full_path}: {[os.path.basename(p) for p in continuation_paths]}")

    for cont_path in continuation_paths:
        try:
            with open(cont_path) as f:
                cont_data = json.load(f)
        except Exception as e:
            print(f"*** ERROR ***: Failed to load continuation file {cont_path} for base {full_path}: {e}")
            continue

        if len(cont_data) != len(data):
            print(f"*** WARNING ***: Continuation file {cont_path} has {len(cont_data)} run(s) but base file {full_path} has {len(data)}; merging only overlapping runs by position.")

        for run_idx, cont_run in enumerate(cont_data):
            if run_idx >= len(data):
                print(f"*** WARNING ***: Continuation file {cont_path} has an extra run at index {run_idx} with no matching run in base file {full_path}; skipping it.")
                continue
            base_run = data[run_idx]
            base_ft_by_seed = {ft_entry.get("seed"): ft_entry for ft_entry in base_run.get("ft", [])}
            for cont_ft in cont_run.get("ft", []):
                seed = cont_ft.get("seed")
                n_new = len(cont_ft.get("stats", []))
                if seed in base_ft_by_seed:
                    base_ft_by_seed[seed].setdefault("stats", []).extend(cont_ft.get("stats", []))
                    print(f"Merged {n_new} continuation stats entries from {os.path.basename(cont_path)} into seed {seed} (run {run_idx}) of {os.path.basename(full_path)}")
                else:
                    print(f"*** WARNING ***: Seed {seed} from {os.path.basename(cont_path)} not found among base seeds {list(base_ft_by_seed.keys())} in {os.path.basename(full_path)}; appending as a new 'ft' entry instead of merging.")
                    base_run.setdefault("ft", []).append(cont_ft)
                    base_ft_by_seed[seed] = cont_ft
    return data

def lists_extend(lists):
    master_list = []
    for lst in lists:
        master_list.extend(lst)
    return master_list

def element_wise_mean(lists):
    max_length = max(len(lst) for lst in lists)
    
    padded_lists = [lst + [float('nan')] * (max_length - len(lst)) for lst in lists]
    
    df = pd.DataFrame(padded_lists)
    mean_values = df.mean(skipna=True).tolist()
    
    return mean_values

def element_wise_std(lists):
    max_length = max(len(lst) for lst in lists)

    padded_lists = [lst + [float('nan')] * (max_length - len(lst)) for lst in lists]

    df = pd.DataFrame(padded_lists)
    std_values = df.std(skipna=True).tolist()
    
    return std_values

def element_wise_diff(list1, list2):
    assert len(list1) == len(list2), "Lists must be of the same length"
    diff_list = []
    for a, b in zip(list1, list2):
        if a is None or b is None:
            diff_list.append(None)
        else:
            diff_list.append(a - b)
    return diff_list

def element_wise_sum(list1, list2):
    assert len(list1) == len(list2), "Lists must be of the same length"
    sum_list = []
    for a, b in zip(list1, list2):
        if a is None or b is None:
            sum_list.append(None)
        else:
            sum_list.append(a + b)
    return sum_list

def get_custom_column(row, column_name):
    label = row["label"]
    df_stats = pd.DataFrame(row["stats"])
    if column_name not in df_stats.columns:
        print(f"*** Warning ***: Column {column_name} not found in stats for label {label}.")
        return []
    df_stats = df_stats.dropna(subset=[column_name], ignore_index=True)
    df_stats = df_stats[(~df_stats["layer_sub"].isna())].head(24).sort_values("layer_sub")
    return_list = df_stats[column_name].to_list()
    if len(return_list) != 24:
        if len(return_list) == 48:
            return_list = return_list[::2]
    return return_list

def get_epoch_column(row, column_name, epoch=299):
    label = row["label"]
    df_stats = pd.DataFrame(row["stats"])
    df_stats = df_stats.dropna(subset=[column_name], ignore_index=True)
    df_stats = df_stats[(~df_stats["layer_sub"].isna())&(df_stats["epoch"] == epoch)].head(24).sort_values("layer_sub")
    if len(df_stats.dropna(subset=[column_name])) == 0 and epoch==299:
        print(f"*** Warning ***: Column {column_name} not found in stats for label {label} at epoch {epoch}. Trying epoch -1 instead.")
        df_stats = df_stats[(~df_stats["layer_sub"].isna())&(df_stats["epoch"] == -1)].head(24).sort_values("layer_sub")

    if column_name not in df_stats.columns:
        print(f"*** Warning ***: Column {column_name} not found in stats for label {label} at epoch {epoch}.")
        return []
    return_list = df_stats[column_name].to_list()
    if len(return_list) != 24:
        if len(return_list) == 48:
            return_list = return_list[::2]
    return return_list
```

# Accuracy processing

```python
def get_layer_accuracies(row, epoch=299):
    """Layer-wise accuracies at a given epoch, coalescing 'acc' with 'attn_acc'/'blk_acc'
    when the unified 'acc' column isn't present. `epoch` defaults to 299 (post-training)
    to preserve existing call sites, but any logged epoch value can be passed."""
    label = row["label"]
    tag = row.get("tag", "No tag")
    path = row.get("path", "Unknown path")
    seed = row.get("seed", "Unknown seed")
    df_stats = pd.DataFrame(row["stats"])
    if "epoch" not in df_stats.columns or "layer_sub" not in df_stats.columns:
        print(f"*** ERROR ***: Missing 'epoch' or 'layer_sub' column for label={label!r}, tag={tag!r}, seed={seed!r}, path={path!r}. Columns found: {df_stats.columns.tolist()}")
    df_stats = df_stats[(~df_stats["layer_sub"].isna())&(df_stats["epoch"] == epoch)].head(24)
    if df_stats.empty:
        available_epochs = pd.DataFrame(row["stats"])["epoch"].unique().tolist() if "epoch" in pd.DataFrame(row["stats"]).columns else "N/A"
        print(f"*** ERROR ***: No rows with epoch=={epoch} for label={label!r}, tag={tag!r}, seed={seed!r}, path={path!r}. Available epochs: {available_epochs}")
    if "acc" in df_stats.columns:
        return_list = df_stats["acc"].to_list()
    else:
        if "attn_acc" not in df_stats.columns or "blk_acc" not in df_stats.columns:
            print(f"*** ERROR ***: No 'acc', 'attn_acc', or 'blk_acc' column for label={label!r}, tag={tag!r}, seed={seed!r}, path={path!r}. Columns found: {df_stats.columns.tolist()}")
        df_stats["acc"] = df_stats.apply(lambda x: x["attn_acc"] if not pd.isna(x["attn_acc"]) else x["blk_acc"], axis=1)
        return_list = df_stats["acc"].to_list()

    if len(return_list) != 24:
        print(f"Warning: Expected 24 accuracies, but got {len(return_list)} for label={label!r}, tag={tag!r}, seed={seed!r}, path={path!r}.")
        if len(return_list) == 48:
            print("Processing every other element to get 24")
            return_list = return_list[::2]
    if len(return_list) == 0:
        print(f"*** ERROR ***: layer_accuracies is EMPTY for label={label!r}, tag={tag!r}, seed={seed!r}, path={path!r}. This will cause an IndexError downstream when indexing [-1].")
    return return_list
```

```python
def process_jsons(json_map):
    all_data = []
    for json_item in json_map:
        json_path = json_item.get("path", "Unknown path")
        print(f"Processing {json_path}...")
        full_path = os.path.join(parent_path, json_path)
        if not os.path.exists(full_path):
            print(f"*** ERROR ***: File not found: {full_path} (from json_map entry: {json_item})")
            continue
        try:
            data = load_json_with_continuations(full_path)
        except Exception as e:
            print(f"*** ERROR ***: Failed to load/parse {full_path} (from json_map entry: {json_item}): {e}")
            continue
        if not data:
            print(f"*** ERROR ***: {full_path} loaded but is empty (from json_map entry: {json_item})")
        for item in data:
            item["category"] = json_item.get("category", "")
            item["tag"] = json_item.get("tag", "")
            item["init"] = json_item.get("init", "")
            item["layer"] = json_item.get("layer", "")
            item["path"] = json_path
        all_data.extend(data)
    if not all_data:
        raise ValueError(f"*** ERROR ***: No data was loaded for any file in json_map: {[j.get('path') for j in json_map]}")
    df = pd.DataFrame(all_data)
    json_map = pd.DataFrame(json_map)

    df = df.explode("ft").reset_index(drop=True).dropna(subset=["ft"], ignore_index=True)
    if df.empty:
        raise ValueError("*** ERROR ***: DataFrame is empty after exploding/dropping NaN 'ft' entries. Check that the source JSON files contain non-null 'ft' data.")
    df["stats"] = df["ft"].apply(lambda x: x["stats"])

    df["label"] = df.apply(lambda x: f"{x['init']}: {x['category']}: {x['layer']}: {x['tag']}", axis=1)
    df["seed"] = df["ft"].apply(lambda x: x["seed"])

    df_columns = df.columns.tolist()
    df = df.groupby(["label", "seed"]).agg({
        "stats": lists_extend,
        **{col: "first" for col in df_columns if col not in ["stats", "label", "seed"]}
    }).reset_index()

    df["layer_accuracies"] = df.apply(lambda x: get_layer_accuracies(x), axis=1)

    df_grouped = df.groupby(["category", "tag", "init", "layer"]).agg({
        "layer_accuracies": element_wise_mean,
        "label": "first",
        "seed": "count"
        }).reset_index()
    df_grouped["layer_accuracies_std"] = df.groupby(["category", "tag", "init", "layer"]).agg({
        "layer_accuracies": element_wise_std,
        }).reset_index()["layer_accuracies"]

    empty_mask = df_grouped["layer_accuracies"].apply(lambda x: len(x) == 0)
    if empty_mask.any():
        bad_rows = df_grouped.loc[empty_mask, ["label", "category", "tag", "init", "layer"]]
        raise IndexError(
            f"*** ERROR ***: 'layer_accuracies' is empty (cannot index [-1]) for the following group(s):\n{bad_rows.to_string(index=False)}\n"
            f"Check the source JSON file(s) for these labels/paths for missing 'acc'/'attn_acc'/'blk_acc' data or no epoch==299 entries (see *** ERROR *** messages above from get_layer_accuracies)."
        )
    df_grouped["final_accuracies"] = df_grouped["layer_accuracies"].apply(lambda x: x[-1])
    df_grouped["final_accuracies_std"] = df_grouped["layer_accuracies_std"].apply(lambda x: x[-1])
    df_grouped["final_accuracies_display"] = df_grouped.apply(lambda x: f"{x['final_accuracies']:.2f} \u00b1 {x['final_accuracies_std']:.2f}", axis=1)
    df_grouped[["category", "label", "final_accuracies_display"]]

    if any(df_grouped["seed"] != 3):
        print("Warning: Some groups do not have 3 seeds. Here are those rows:")
        print(df_grouped[df_grouped["seed"] != 3])
    return df_grouped
```

## delta norm ratio across epochs (gradient)

For each run in `json_map_main`, plots the init (`custom_delta_norm_ratio`) checkpoint as a dashed black line, plus the delta-norm-ratio at each epoch in `epoch_gradient_list` as a solid line, colored on a viridis gradient by epoch.

```python
import matplotlib.cm as _cm
from matplotlib import colors as _colors
from matplotlib.lines import Line2D

_ORDER_PRINT_EPOCH_GRADIENT = ["0-attn", "0-mlp", "1-attn", "1-mlp", "2-attn", "2-mlp", "3-attn", "3-mlp",
                               "4-attn", "4-mlp", "5-attn", "5-mlp", "6-attn", "6-mlp", "7-attn", "7-mlp",
                               "8-attn", "8-mlp", "9-attn", "9-mlp", "10-attn", "10-mlp", "11-attn", "11-mlp"]

_ATTN_SHADE_COLOR = "#FFFFFF"
_MLP_SHADE_COLOR = "#888888"

def process_jsons_norm_epochs(json_map, epochs, compute_final_accuracy=False):
    """Like process_jsons_norm_checkpoints, but pulls delta-norm-ratio at an arbitrary
    list of epochs (plus the "init" custom_delta_norm_ratio snapshot) per
    (init, category, tag, layer) group. Pass compute_final_accuracy=True to also compute
    a "final_accuracy" column (the last-layer accuracy at the last epoch in `epochs`,
    via get_layer_accuracies), used by _plot_epoch_gradient_multi's color_by="accuracy"."""
    all_data = []
    for json_item in json_map:
        full_path = os.path.join(parent_path, json_item["path"])
        data = load_json_with_continuations(full_path)
        for item in data:
            item["category"] = json_item.get("category", "")
            item["tag"] = json_item.get("tag", "")
            item["init"] = json_item.get("init", "")
            item["layer"] = json_item.get("layer", "")
        all_data.extend(data)
    df = pd.DataFrame(all_data)

    df = df.explode("ft").reset_index(drop=True)
    df["stats"] = df["ft"].apply(lambda x: x["stats"])
    df["label"] = df.apply(lambda x: f"{x['init']}: {x['category']}: {x['layer']}: {x['tag']}", axis=1)
    df["seed"] = df["ft"].apply(lambda x: x["seed"])

    df_columns = df.columns.tolist()
    df = df.groupby(["label", "seed"]).agg({
        "stats": lists_extend,
        **{col: "first" for col in df_columns if col not in ["stats", "label", "seed"]}
    }).reset_index()

    df["custom_delta_norm_ratio"] = df.apply(lambda x: get_custom_column(x, "custom_delta_norm_ratio"), axis=1)
    epoch_columns = []
    for epoch in epochs:
        column_name = f"e{epoch}_delta_norm_ratio"
        epoch_columns.append(column_name)
        df[column_name] = df.apply(lambda x: get_epoch_column(x, "delta_norm_ratio", epoch=epoch), axis=1)

    checkpoint_columns = ["custom_delta_norm_ratio"] + epoch_columns

    if compute_final_accuracy:
        final_epoch = max(epochs)
        df["final_accuracy"] = df.apply(lambda x: get_layer_accuracies(x, epoch=final_epoch)[-1:], axis=1)
        checkpoint_columns = checkpoint_columns + ["final_accuracy"]

    df_grouped = df.groupby(["init", "category", "tag", "layer"]).agg({
        "label": "first",
        "seed": "count",
    }).reset_index()

    for column_name in checkpoint_columns:
        df_grouped[column_name] = df.groupby(["label"]).agg({column_name: element_wise_mean}).reset_index()[column_name]
        df_grouped[f"{column_name}_std"] = df.groupby(["label"]).agg({column_name: element_wise_std}).reset_index()[column_name]

    if any(df_grouped["seed"] != 3):
        print("Warning: Some groups do not have 3 seeds. Here are those rows:")
        print(df_grouped[df_grouped["seed"] != 3])

    return df_grouped

def _add_layer_shading(ax):
    """Light background bands behind each x-position: one color for attn columns,
    another for mlp columns, so attn/mlp stay visually distinguishable even though
    the tick labels only show the layer number."""
    for i, lbl in enumerate(_ORDER_PRINT_EPOCH_GRADIENT):
        color = _ATTN_SHADE_COLOR if lbl.endswith("attn") else _MLP_SHADE_COLOR
        ax.axvspan(i - 0.5, i + 0.5, color=color, alpha=0.1, zorder=0, lw=0)

def _project_series(values, mode):
    """Maps a delta-norm-ratio series (right-aligned to the tail of
    _ORDER_PRINT_EPOCH_GRADIENT) to (x, y) pairs for plotting. mode="all" keeps every
    attn/mlp position as-is (x runs over the full 24-slot index). mode="attn"/"mlp"
    keeps only the matching type, positioned at its layer number (x runs 0..11), since
    only half the positions apply and the interleaved indices would otherwise leave gaps."""
    if mode == "all":
        n_positions = len(_ORDER_PRINT_EPOCH_GRADIENT)
        x = list(range(n_positions - len(values), n_positions))
        return x, list(values)

    labels_slice = _ORDER_PRINT_EPOCH_GRADIENT[-len(values):]
    pairs = [(int(lbl.split("-")[0]), v) for lbl, v in zip(labels_slice, values) if lbl.endswith(mode)]
    if not pairs:
        return [], []
    xs, ys = zip(*pairs)
    return list(xs), list(ys)

def _plot_line_with_markers(ax, x, values, color, linestyle, label, mode="all"):
    """Plots one line. In mode="all", attn points get a circle marker and mlp points
    get a triangle marker, using the "N-attn"/"N-mlp" labels sliced to match `x` to
    tell them apart. In mode="attn"/"mlp" every point is the same type, so it's drawn
    with a single plain scatter."""
    ax.plot(x, values, linestyle=linestyle, linewidth=1, color=color, label=label, zorder=2)

    if mode == "all":
        labels_slice = _ORDER_PRINT_EPOCH_GRADIENT[-len(values):]
        is_attn = [lbl.endswith("attn") for lbl in labels_slice]

        attn_x = [xi for xi, a in zip(x, is_attn) if a]
        attn_y = [yi for yi, a in zip(values, is_attn) if a]
        mlp_x = [xi for xi, a in zip(x, is_attn) if not a]
        mlp_y = [yi for yi, a in zip(values, is_attn) if not a]

        ax.scatter(attn_x, attn_y, s=2, color=color, zorder=3)
        ax.scatter(mlp_x, mlp_y, s=2, color=color, zorder=3)
    else:
        ax.scatter(x, values, s=2, color=color, zorder=3)

def _set_epoch_gradient_xaxis(ax, mode):
    n_positions = len(_ORDER_PRINT_EPOCH_GRADIENT)
    n_layers = n_positions // 2
    if mode == "all":
        ax.set_xlim(-0.5, n_positions - 0.5)
        tick_positions = [i + 0.5 for i in range(0, n_positions, 2)]
    else:
        ax.set_xlim(-0.5, n_layers - 0.5)
        tick_positions = list(range(n_layers))
    tick_labels = [str(i) for i in range(n_layers)]
    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_labels, rotation=0)

def _plot_epoch_gradient(row, epochs, file_name_string, set_name="", plot_init=False, mode="all"):
    fig, ax = plt.subplots()

    cmap = plt.get_cmap("Greys")
    norm = _colors.Normalize(vmin=-20, vmax=max(epochs)) if epochs else None

    if mode == "all":
        _add_layer_shading(ax)

    init_values = row["custom_delta_norm_ratio"] if plot_init else None
    if isinstance(init_values, list) and len(init_values) > 0:
        init_std = row["custom_delta_norm_ratio_std"]
        x, y = _project_series(init_values, mode)
        _, y_std = _project_series(init_std, mode)
        if y:
            _plot_line_with_markers(ax, x, y, color='black', linestyle='-', label='Init', mode=mode)
            ax.fill_between(x, element_wise_diff(y, y_std), element_wise_sum(y, y_std), alpha=0.15, color='black', zorder=1)
    else:
        print(f"Skipping Init for {row['label']}: no data in column custom_delta_norm_ratio")

    for epoch in epochs:
        column_name = f"e{epoch}_delta_norm_ratio"
        values = row[column_name]
        if not isinstance(values, list) or len(values) == 0:
            print(f"Skipping epoch {epoch} for {row['label']}: no data in column {column_name}")
            continue
        std = row[f"{column_name}_std"]
        x, y = _project_series(values, mode)
        _, y_std = _project_series(std, mode)
        if not y:
            print(f"Skipping epoch {epoch} for {row['label']}: no {mode} data")
            continue
        color = cmap(norm(epoch))
        _plot_line_with_markers(ax, x, y, color=color, linestyle='-', label=f"Epoch {epoch}", mode=mode)
        ax.fill_between(x, element_wise_diff(y, y_std), element_wise_sum(y, y_std), alpha=0.2, color=color, zorder=1)

    sm = _cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])

    _set_epoch_gradient_xaxis(ax, mode)

    ax.set_xlabel('Layer')
    ax.set_ylabel( "Residual write ratio", rotation=90)
    ax.set_yscale('log')
    ax.set_ylim(0.01, 5)
    plt.grid(alpha=0.3)

    save_name = f"{set_name}_{file_name_string}".strip("_")
    plt.savefig(f"{OUTPUT_DIR_PDF}/{save_name}.pdf", format="pdf", bbox_inches='tight', dpi=1500)
    plt.savefig(f"{OUTPUT_DIR_PNG}/{save_name}.png", format="png", bbox_inches='tight', dpi=1500)
    plt.show()

def plot_norm_epoch_gradient(json_map, epochs, set_name_="", plot_init=False):
    """For each dict in json_map (same shape as json_map_main entries: init/category/
    layer/tag/path), plot delta-norm-ratio across layers as one figure with a dashed
    "Init" line plus one line per epoch in `epochs`, colored on a viridis gradient by
    epoch. Pass plot_init=True to include the "Init" line. For each run this saves
    three figures: the combined attn+mlp plot, an attention-only plot (filename
    suffix "_attn"), and an mlp-only plot (filename suffix "_mlp")."""
    for json_item in json_map:
        df_grouped = process_jsons_norm_epochs([json_item], epochs)
        if df_grouped.empty:
            print(f"No data found for {json_item}")
            continue
        row = df_grouped.iloc[0]

        file_name_base = set_name_+f"{json_item.get('init','')}_{json_item.get('category','')}_{json_item.get('layer','')}_{json_item.get('tag','')}"
        file_name_base = file_name_base.replace("/", "_").replace(" ", "_").replace("-", "_").replace("__", "_").strip("_")

        for mode, suffix in (("all", ""), ("attn", "_attn"), ("mlp", "_mlp")):
            _plot_epoch_gradient(row, epochs, f"{file_name_base}_delta_norm_ratio_epoch_gradient{suffix}", set_name, plot_init=plot_init, mode=mode)

_MULTI_RUN_CMAPS = ["Blues", "Oranges", "Greens", "Purples", "Reds", "Greys"]
_MULTI_RUN_TAB10 = sns.color_palette("tab10")

def _plot_epoch_gradient_multi(rows, epochs, file_name_string, color_by="epoch", plot_init=True, mode="all", show_legend=True):
    """Like _plot_epoch_gradient, but overlays multiple runs (rows) in one figure.
    color_by="epoch" (default) gives each run its own sequential colormap family (so it
    stays visually distinct from the other runs) shaded light-to-dark by epoch within
    that family. color_by="tab10" instead gives each run one flat color from the
    seaborn tab10 palette (same color for every epoch/init line within a run).
    color_by="accuracy" gives each run one flat color from a single shared viridis
    colormap, positioned by that run's "final_accuracy" column (its last-layer accuracy
    at the last epoch in `epochs`; rows must come from process_jsons_norm_epochs called
    with compute_final_accuracy=True) - every line in a run shares that run's color, and
    a colorbar labeled "Final accuracy" is added. The legend shows one entry per run
    using its 'init' field. Pass show_legend=False to omit it."""
    fig, ax = plt.subplots()

    if mode == "all":
        _add_layer_shading(ax)

    norm = _colors.Normalize(vmin=min(epochs), vmax=max(epochs)) if epochs else None
    legend_handles = []

    accuracy_cmap = None
    accuracy_norm = None
    if color_by == "accuracy":
        accuracy_cmap = plt.get_cmap("viridis")
        final_accuracies = [row["final_accuracy"][0] for row in rows if isinstance(row.get("final_accuracy"), list) and len(row["final_accuracy"]) > 0]
        if not final_accuracies:
            raise ValueError("color_by=\"accuracy\" requires a non-empty 'final_accuracy' column; pass compute_final_accuracy=True to process_jsons_norm_epochs.")
        accuracy_norm = _colors.Normalize(vmin=min(final_accuracies), vmax=max(final_accuracies))

    for run_idx, row in enumerate(rows):
        if color_by == "tab10":
            flat_color = _MULTI_RUN_TAB10[run_idx % len(_MULTI_RUN_TAB10)]
        elif color_by == "accuracy":
            flat_color = accuracy_cmap(accuracy_norm(row["final_accuracy"][0]))
        else:
            cmap = plt.get_cmap(_MULTI_RUN_CMAPS[run_idx % len(_MULTI_RUN_CMAPS)])
        run_label = row["init"]

        init_values = row["custom_delta_norm_ratio"] if plot_init else None
        if isinstance(init_values, list) and len(init_values) > 0:
            init_std = row["custom_delta_norm_ratio_std"]
            x, y = _project_series(init_values, mode)
            _, y_std = _project_series(init_std, mode)
            if y:
                init_color = flat_color if color_by in ("tab10", "accuracy") else cmap(0.8)
                _plot_line_with_markers(ax, x, y, color=init_color, linestyle='-', label=None, mode=mode)
                ax.fill_between(x, element_wise_diff(y, y_std), element_wise_sum(y, y_std), alpha=0.15, color=init_color, zorder=1)
        else:
            print(f"Skipping Init for {row['label']}: no data in column custom_delta_norm_ratio")

        for epoch in epochs:
            column_name = f"e{epoch}_delta_norm_ratio"
            values = row[column_name]
            if not isinstance(values, list) or len(values) == 0:
                print(f"Skipping epoch {epoch} for {row['label']}: no data in column {column_name}")
                continue
            std = row[f"{column_name}_std"]
            x, y = _project_series(values, mode)
            _, y_std = _project_series(std, mode)
            if not y:
                print(f"Skipping epoch {epoch} for {row['label']}: no {mode} data")
                continue
            color = flat_color if color_by in ("tab10", "accuracy") else cmap(norm(epoch))
            _plot_line_with_markers(ax, x, y, color=color, linestyle='-', label=None, mode=mode)
            ax.fill_between(x, element_wise_diff(y, y_std), element_wise_sum(y, y_std), alpha=0.15, color=color, zorder=1)

        legend_handles.append(Line2D([0], [0], color=(flat_color if color_by in ("tab10", "accuracy") else cmap(0.7)), lw=2, label=run_label))

    _set_epoch_gradient_xaxis(ax, mode)

    ax.set_xlabel('Layer')
    ax.set_ylabel( "Residual write ratio", rotation=90)
    ax.set_yscale('log')
    if color_by == "accuracy":
        sm = _cm.ScalarMappable(cmap=accuracy_cmap, norm=accuracy_norm)
        sm.set_array([])
        fig.colorbar(sm, ax=ax, label="Final accuracy")
    if show_legend:
        ax.legend(handles=legend_handles, loc='lower center', bbox_to_anchor=(0.5, 1.02), ncol=len(legend_handles), frameon=False)
    plt.grid(alpha=0.3)

    save_name = f"{file_name_string}".strip("_")
    plt.savefig(f"{OUTPUT_DIR_PDF}/{save_name}.pdf", format="pdf", bbox_inches='tight', dpi=1500)
    plt.savefig(f"{OUTPUT_DIR_PNG}/{save_name}.png", format="png", bbox_inches='tight', dpi=1500)
    print(f"Saved multi-run epoch gradient plot to {OUTPUT_DIR_PDF}/{save_name}.pdf and {OUTPUT_DIR_PNG}/{save_name}.png")
    plt.show()

def plot_norm_epoch_gradient_multi(json_map, epochs, set_prefix="", color_by="epoch", plot_init=True, show_legend=True):
    """Like plot_norm_epoch_gradient, but draws all runs in json_map on a single
    shared figure instead of one figure per run. color_by selects how runs are colored:
    "epoch" (default) uses a per-run sequential colormap family shaded by epoch,
    "tab10" gives each run one flat seaborn tab10 color, and "accuracy" gives each run
    one flat color from a shared viridis colormap positioned by that run's final
    accuracy (with a "Final accuracy" colorbar). Pass plot_init=False to omit the
    "Init" line. Pass show_legend=False to omit the per-run legend above the plot.
    Saves three figures: the combined attn+mlp plot, an attention-only plot (filename
    suffix "_attn"), and an mlp-only plot (filename suffix "_mlp")."""
    rows = []
    for json_item in json_map:
        df_grouped = process_jsons_norm_epochs([json_item], epochs, compute_final_accuracy=(color_by == "accuracy"))
        if df_grouped.empty:
            print(f"No data found for {json_item}")
            continue
        rows.append(df_grouped.iloc[0])

    if len(rows) == 0:
        print("No data found for any run in json_map")
        return

    file_name_base = set_prefix+"_"+set_name

    epoch_str = "_".join(f"e{epoch}" for epoch in epochs)

    for mode, mode_suffix in (("all", ""), ("attn", "_attn"), ("mlp", "_mlp")):
        save_label_parts = [file_name_base, f"delta_norm_ratio_epoch_gradient_multi{mode_suffix}", epoch_str]
        if plot_init:
            save_label_parts.append("init")
        if color_by == "accuracy":
            save_label_parts.append("acccolor")
        save_label = "_".join(part for part in save_label_parts if part)

        _plot_epoch_gradient_multi(rows, epochs, save_label, color_by=color_by, plot_init=plot_init, mode=mode, show_legend=show_legend)
```
